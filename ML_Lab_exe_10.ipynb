```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression

urls = ["google.com", "youtube.com", "github.com",
        "login-secure.xyz", "free-prize.tk"]

labels = ["Safe", "Safe", "Safe", "Malicious", "Malicious"]

vectorizer = TfidfVectorizer(analyzer="char", ngram_range=(2, 5))
X = vectorizer.fit_transform(urls)

model = LogisticRegression()
model.fit(X, labels)

url = input("Enter URL: ")
result = model.predict(vectorizer.transform([url]))[0]

print("URL:", url)
print("Classification:", result)
```

```python
url = "login-secure.xyz"

result = model.predict(
    vectorizer.transform([url])
)[0]

print("URL:", url)
print("Classification:", result)

if result == "Malicious":
    print("WARNING: Suspicious URL detected!")
```

```python
test_urls = [
    "google.com",
    "youtube.com",
    "free-prize.tk",
    "github.com",
    "bank-login.xyz"
]

for url in test_urls:
    result = model.predict(
        vectorizer.transform([url])
    )[0]

    print(url, " --> ", result)
```

```python
url = "http://login-secure.xyz"

print("URL:", url)
print("Length:", len(url))
print("Dots:", url.count("."))
print("Hyphens:", url.count("-"))
print("HTTPS:", url.startswith("https"))
```

```python
url = input("Enter Website URL: ")

result = model.predict(
    vectorizer.transform([url])
)[0]

print("\n===== WebShield Intelligence =====")
print("Website:", url)
print("Result:", result)

if result == "Safe":
    print("Status: Website appears safe.")
else:
    print("Status: Malicious URL detected!")
    print("Action: Avoid opening this website.")
```
